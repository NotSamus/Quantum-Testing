```python
import qiskit
```

```python
from qiskit import QuantumCircuit  # Import quantum circuit module
```

```python
# This is the third bell state
qc1 = QuantumCircuit(2)

qc1.x(1)
qc1.h(0)
qc1.cx(0,1)

qc1.draw(output='mpl')
```

```python
# This is the fourth bell state as circuit
qc = QuantumCircuit(2)



qc.x(1)
qc.x(0)
qc.h(0)
qc.cx(0,1)
qc.draw(output='mpl')  # Draw the circuit diagram
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

def get_state_equation(qc):
    """Returns the quantum state equation from a circuit."""
    state = Statevector.from_instruction(qc)  # Get statevector
    return state 
```

```python
quantum_state= get_state_equation(qc)
```

```python
print(quantum_state)
```

```python
import numpy as np

def pretty_print_state(state):
    """Formats a statevector as a readable quantum equation."""
    basis_states = ["|00⟩", "|01⟩", "|10⟩", "|11⟩"]
    equation = " + ".join(
        f"({np.round(c.real, 3)} + {np.round(c.imag, 3)}i) {basis_states[i]}"
        for i, c in enumerate(state.data)
        if not np.isclose(c, 0)  # Ignore zero amplitudes
    )
    return equation

# Print the quantum equation
print(pretty_print_state(quantum_state))
```
